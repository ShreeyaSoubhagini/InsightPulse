```python
# Prepare review text for NLP

import re
import string

def clean_review(text):
    text = text.lower()
    text = re.sub(r"http\S+|www\S+", "", text)
    text = text.translate(str.maketrans("", "", string.punctuation))
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["Clean_Review"] = df["Review Text"].apply(clean_review)

df[["Review Text", "Clean_Review"]].head(10)
```

```python
# most frequent words

from collections import Counter
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

all_words = " ".join(df["Clean_Review"]).split()

filtered_words = [
    word for word in all_words
    if word not in ENGLISH_STOP_WORDS
    and len(word) > 2
]

word_counts = Counter(filtered_words)

top_words = word_counts.most_common(20)

top_words
```

```python
# Compare words in positive vs negative reviews

positive_text = " ".join(
    df.loc[df["Rating"] >= 4, "Clean_Review"]
)

negative_text = " ".join(
    df.loc[df["Rating"] <= 2, "Clean_Review"]
)

positive_words = [
    word for word in positive_text.split()
    if word not in ENGLISH_STOP_WORDS
    and len(word) > 2
]

negative_words = [
    word for word in negative_text.split()
    if word not in ENGLISH_STOP_WORDS
    and len(word) > 2
]

positive_counts = Counter(positive_words)
negative_counts = Counter(negative_words)

print("Top Positive Review Words:")
print(positive_counts.most_common(20))

print("\nTop Negative Review Words:")
print(negative_counts.most_common(20))
```

```python
# Normalize word frequency by review group

positive_reviews = df[df["Rating"] >= 4]["Clean_Review"]
negative_reviews = df[df["Rating"] <= 2]["Clean_Review"]

positive_word_presence = Counter()
negative_word_presence = Counter()

for review in positive_reviews:
    words = set(
        word for word in review.split()
        if word not in ENGLISH_STOP_WORDS and len(word) > 2
    )
    positive_word_presence.update(words)

for review in negative_reviews:
    words = set(
        word for word in review.split()
        if word not in ENGLISH_STOP_WORDS and len(word) > 2
    )
    negative_word_presence.update(words)

positive_total = len(positive_reviews)
negative_total = len(negative_reviews)

word_comparison = []

all_words_set = set(positive_word_presence) | set(negative_word_presence)

for word in all_words_set:
    positive_pct = (
        positive_word_presence[word] / positive_total * 100
    )
    
    negative_pct = (
        negative_word_presence[word] / negative_total * 100
    )
    
    word_comparison.append(
        (
            word,
            positive_pct,
            negative_pct,
            negative_pct - positive_pct
        )
    )

word_comparison_df = pd.DataFrame(
    word_comparison,
    columns=[
        "Word",
        "Positive_Percentage",
        "Negative_Percentage",
        "Difference"
    ]
)

word_comparison_df = word_comparison_df.sort_values(
    "Difference",
    ascending=False
)

word_comparison_df.head(20).round(2)
```

```python
# Identify complaint themes

# Define complaint themes using keyword groups

theme_keywords = {
    "Fit & Size": [
        "size", "fit", "small", "large", "tight", "loose",
        "oversized", "length", "short", "long", "waist"
    ],
    
    "Fabric & Material": [
        "fabric", "material", "quality", "thin", "thick",
        "cheap", "scratchy", "rough", "soft", "texture"
    ],
    
    "Appearance & Style": [
        "color", "colour", "look", "looks", "style",
        "design", "flattering", "unflattering", "pattern"
    ],
    
    "Comfort": [
        "comfortable", "comfort", "uncomfortable",
        "scratchy", "itchy", "soft"
    ],
    
    "Returns & Exchanges": [
        "return", "returned", "returning",
        "refund", "exchange"
    ],
    
    "Quality & Durability": [
        "quality", "broke", "broken", "tear", "torn",
        "damage", "damaged", "stitch", "seam"
    ]
}
```

```python
negative_reviews_df = df[
    (df["Rating"] <= 2) &
    (df["Review Text"] != "No Review")
].copy()

theme_results = []

for theme, keywords in theme_keywords.items():
    
    count = 0
    
    for review in negative_reviews_df["Clean_Review"]:
        words = set(review.split())
        
        if any(keyword in words for keyword in keywords):
            count += 1
    
    percentage = count / len(negative_reviews_df) * 100
    
    theme_results.append(
        [theme, count, percentage]
    )

theme_analysis = pd.DataFrame(
    theme_results,
    columns=[
        "Theme",
        "Negative_Review_Count",
        "Percentage_of_Negative_Reviews"
    ]
)

theme_analysis = theme_analysis.sort_values(
    "Negative_Review_Count",
    ascending=False
)

theme_analysis.round(2)
```

```python
# Which themes are associated with the lowest ratings?

theme_rating_results = []

for theme, keywords in theme_keywords.items():
    
    mask = df["Clean_Review"].apply(
        lambda text: any(keyword in set(text.split()) for keyword in keywords)
    )
    
    theme_reviews = df[mask]
    
    theme_rating_results.append([
        theme,
        len(theme_reviews),
        theme_reviews["Rating"].mean(),
        theme_reviews["Recommended IND"].mean() * 100
    ])

theme_rating_analysis = pd.DataFrame(
    theme_rating_results,
    columns=[
        "Theme",
        "Review_Count",
        "Average_Rating",
        "Recommendation_Rate"
    ]
)

theme_rating_analysis = theme_rating_analysis.sort_values(
    "Average_Rating"
)

theme_rating_analysis.round(2)
```

```python
# Let's connect complaints to product categories

important_themes = [
    "Fit & Size",
    "Appearance & Style",
    "Fabric & Material"
]

theme_class_results = []

for class_name, group in df.groupby("Class Name"):
    
    result = {
        "Class Name": class_name,
        "Review_Count": len(group)
    }
    
    for theme in important_themes:
        keywords = theme_keywords[theme]
        
        count = 0
        
        for review in group["Clean_Review"]:
            words = set(review.split())
            
            if any(keyword in words for keyword in keywords):
                count += 1
        
        result[theme] = count / len(group) * 100
    
    theme_class_results.append(result)

theme_class_analysis = pd.DataFrame(theme_class_results)

# Keep only meaningful classes with at least 100 reviews
theme_class_analysis = theme_class_analysis[
    theme_class_analysis["Review_Count"] >= 100
]

theme_class_analysis.round(2)
```

```python
# SQL: Find the highest-rated departments with enough reviews

query = """
SELECT
    "Department Name",
    COUNT(*) AS Review_Count,
    ROUND(AVG(Rating), 2) AS Average_Rating,
    ROUND(AVG("Recommended IND") * 100, 2) AS Recommendation_Rate
FROM reviews
GROUP BY "Department Name"
HAVING COUNT(*) >= 1000
ORDER BY Average_Rating DESC;
"""

department_filtered_sql = pd.read_sql_query(query, conn)

department_filtered_sql
```

```python
# SQL: Negative reviews by department

query = """
SELECT
    "Department Name",
    COUNT(*) AS Review_Count,
    SUM(CASE WHEN Rating < 3 THEN 1 ELSE 0 END) AS Negative_Reviews,
    ROUND(
        SUM(CASE WHEN Rating < 3 THEN 1 ELSE 0 END) * 100.0
        / COUNT(*),
        2
    ) AS Negative_Rate
FROM reviews
GROUP BY "Department Name"
HAVING COUNT(*) >= 1000
ORDER BY Negative_Rate DESC;
"""

negative_department_sql = pd.read_sql_query(query, conn)

negative_department_sql
```

```python
# SQL: Rating distribution by department

query = """
SELECT
    "Department Name",
    Rating,
    COUNT(*) AS Review_Count
FROM reviews
WHERE "Department Name" != 'Unknown'
GROUP BY "Department Name", Rating
ORDER BY "Department Name", Rating;
"""

rating_department_sql = pd.read_sql_query(query, conn)

rating_department_sql
```

```python
# SQL: Customer age-group analysis

query = """
SELECT
    CASE
        WHEN Age BETWEEN 18 AND 25 THEN '18-25'
        WHEN Age BETWEEN 26 AND 35 THEN '26-35'
        WHEN Age BETWEEN 36 AND 45 THEN '36-45'
        WHEN Age BETWEEN 46 AND 55 THEN '46-55'
        WHEN Age BETWEEN 56 AND 65 THEN '56-65'
        ELSE '66+'
    END AS Age_Group,

    COUNT(*) AS Review_Count,

    ROUND(AVG(Rating), 2) AS Average_Rating,

    ROUND(
        AVG("Recommended IND") * 100,
        2
    ) AS Recommendation_Rate

FROM reviews

GROUP BY Age_Group

ORDER BY Average_Rating DESC;
"""

age_sql = pd.read_sql_query(query, conn)

age_sql
```

```python
# SQL ranking: Top products by performance

query = """
WITH product_metrics AS (
    SELECT
        "Department Name",
        "Clothing ID",
        COUNT(*) AS Review_Count,
        ROUND(AVG(Rating), 2) AS Average_Rating,
        ROUND(AVG("Recommended IND") * 100, 2) AS Recommendation_Rate
    FROM reviews
    GROUP BY
        "Department Name",
        "Clothing ID"
    HAVING COUNT(*) >= 50
),

ranked_products AS (
    SELECT
        *,
        RANK() OVER (
            PARTITION BY "Department Name"
            ORDER BY Average_Rating DESC
        ) AS Product_Rank
    FROM product_metrics
)

SELECT *
FROM ranked_products
WHERE Product_Rank <= 3
ORDER BY "Department Name", Product_Rank;
"""

top_products_sql = pd.read_sql_query(query, conn)

top_products_sql
```

```python
# Final SQL query

query = """
WITH department_metrics AS (
    SELECT
        "Department Name",
        COUNT(*) AS Review_Count,

        ROUND(AVG(Rating), 2) AS Average_Rating,

        ROUND(
            AVG("Recommended IND") * 100,
            2
        ) AS Recommendation_Rate,

        SUM(
            CASE
                WHEN Rating < 3 THEN 1
                ELSE 0
            END
        ) AS Negative_Reviews

    FROM reviews

    WHERE "Department Name" != 'Unknown'

    GROUP BY "Department Name"
),

final_metrics AS (
    SELECT
        *,
        ROUND(
            Negative_Reviews * 100.0 / Review_Count,
            2
        ) AS Negative_Rate
    FROM department_metrics
)

SELECT
    "Department Name",
    Review_Count,
    Average_Rating,
    Recommendation_Rate,
    Negative_Reviews,
    Negative_Rate
FROM final_metrics
ORDER BY Negative_Reviews DESC;
"""

final_department_sql = pd.read_sql_query(query, conn)

final_department_sql
```
